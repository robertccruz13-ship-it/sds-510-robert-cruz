```python
# Housing Costs, Financial Stress, and Life Transitions Among Gen Z
# Pilot Data Analysis

import pandas as pd
import matplotlib.pyplot as plt


from google.colab import files
uploaded = files.upload()


df = pd.read_csv("genz survey (1).csv")


print("First 5 rows:")
display(df.head())


print("Sample size:", len(df))
print("\nAge Distribution")
print(df["Q1 : What is your age?"].value_counts())

print("\nEmployment Status")
print(df["Q2 : What is your primary employment status?"].value_counts())

print("\nLiving Arrangement")
print(df["Q10: What is your current living arrangement?"].value_counts())


stress_cols = [
    "Q6: In the past 3 months, I have had difficulty paying my bills.",
    "Q7 : In the past 3 months, I have felt anxious about my financial situation.",
    "Q8: In the past 3 months, I have been unable to save money  consistently.",
    "Q9: I worry about my long term financial security."
]


likert_map = {
    "Strongly disagree": 1,
    "Disagree": 2,
    "Neutral": 3,
    "Agree": 4,
    "Strongly agree": 5
}

# Diagnostic step: Print DataFrame columns to identify exact names
print("\nDataFrame Columns:")
print(df.columns)

for col in stress_cols:
    df[col] = df[col].map(likert_map)


df["financial_stress_score"] = df[stress_cols].mean(axis=1)

print("\nFinancial Stress Scores")
print(df["financial_stress_score"])

print("\nAverage Financial Stress Score:")
print(df["financial_stress_score"].mean())


df["financial_stress_score"].plot(kind="bar")
plt.title("Pilot Financial Stress Scores")
plt.xlabel("Respondent")
plt.ylabel("Average Stress Score")
plt.show()
```

# Pilot Data Analysis

This notebook contains preliminary pilot analysis for my survey project examining housing cost burden, financial stress, and delayed adulthood among Gen Z adults in the United States.

The purpose of this pilot is to test survey feasibility, response quality, and variable structure rather than draw substantive conclusions.

## Preliminary Findings

The pilot produced 4 responses, which is sufficient for testing survey structure but not substantive statistical analysis.

Initial responses suggest variation in:
- income
- housing burden

- living arrangements

This indicates that the survey instrument is capable of capturing meaningful variation for a larger future sample.

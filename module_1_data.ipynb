```python
import pandas as pd

url = "https://raw.githubusercontent.com/evangambit/JsonOfCounties/master/counties.json"
df = pd.read_json(url)

df.head()
```

I am using the JsonOfCountries dataset, which contains county level data for the United States. This analysis is individual countries. It includes variables such as population, income, education, and poverty rates. I loaded this dataset from Github and inspected its structure using pandas function.


```python
df['avg_income'].describe()
```

```python
df ['state'].value_counts()
```

I am choosing these 4 variables. Population, income, poverty, and education. Populatiion measures the total number of people in each county. Income represents the median household income and helps compare economic well being. Poverty shows the percentage of people living below the poverty line. Education shows the eduaction levels, such as percentage with college degrees.

```python
import matplotlib.pyplot as plt

df['avg_income'].hist()
plt.title('Distribution of Median Income')
plt.xlabel('Income')
plt.ylabel('Number of Counties')
plt.show()
```

```python
df['bachelors_degree_plus'] = df['edu'].apply(lambda x: x['bachelors+'])
df.plot.scatter(x='bachelors_degree_plus', y='poverty-rate')
plt.title('Poverty Rate vs. Percentage with Bachelor\'s Degree or Higher')
plt.xlabel('Percentage with Bachelor\'s Degree or Higher')
plt.ylabel('Poverty Rate')
plt.show()
```

The first graph shows that most counties hover around 50000 in the median income. In the second graph, there appears to be a negative relationship, the counties with hiogher education levels tend to have lower poverty levels.

```python
df.groupby('state')['avg_income'].mean().sort_values(ascending=False)
```

```python
high_ed = df[df['bachelors_degree_plus'] > df['bachelors_degree_plus'].median()]
low_ed = df[df['bachelors_degree_plus'] < df['bachelors_degree_plus'].median()]
high_ed['poverty-rate'].mean(), low_ed['poverty-rate'].mean()
```

Some states have higher average incomes than other, showing regional inequlity. Counties with higher education levels tend to have lower poverty rates compared to counties with lower education.
